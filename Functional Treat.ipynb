# 📊 FUNCTIONAL TREAT 📊

```python
def input_data():
    data = input("\nEnter data for a 1D array (separated by spaces): ").split()
    return list(map(int, data))


def display_summary(data):
    if len(data) == 0:
        print("\nNo data available! Please input data first.")
        return

    print("\nData Summary:")
    print("- Total elements:", len(data))
    print("- Minimum value:", min(data))
    print("- Maximum value:", max(data))
    print("- Sum of all values:", sum(data))
    print("- Average value:", round(sum(data) / len(data), 2))


def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


def filter_data(data):
    if len(data) == 0:
        print("\nNo data available! Please input data first.")
        return

    threshold = int(input("\nEnter a threshold value to filter out data above this value: "))

    filtered = list(filter(lambda x: x > threshold, data))

    print("\nFiltered Data (values > " + str(threshold) + "):")

    if len(filtered) > 0:
        print(", ".join(map(str, filtered)))
    else:
        print("No values found.")


def sort_data(data):
    if len(data) == 0:
        print("\nNo data available! Please input data first.")
        return

    print("\nChoose sorting option:")
    print("1. Ascending")
    print("2. Descending")

    choice = int(input("\nEnter your choice: "))

    if choice == 1:
        sorted_data = sorted(data)
        print("\nSorted Data in Ascending Order:")
        print(", ".join(map(str, sorted_data)))

    elif choice == 2:
        sorted_data = sorted(data, reverse=True)
        print("\nSorted Data in Descending Order:")
        print(", ".join(map(str, sorted_data)))

    else:
        print("\nInvalid choice!")


def dataset_statistics(data):
    if len(data) == 0:
        print("\nNo data available! Please input data first.")
        return

    minimum = min(data)
    maximum = max(data)
    total = sum(data)
    average = round(total / len(data), 2)

    return minimum, maximum, total, average


data_list = []

print("Welcome to the Data Analyzer and Transformer Program")

while True:
    print("\nMain Menu:")
    print("1. Input Data")
    print("2. Display Data Summary (Built-in Functions)")
    print("3. Calculate Factorial (Recursion)")
    print("4. Filter Data by Threshold (Lambda Function)")
    print("5. Sort Data")
    print("6. Display Dataset Statistics (Return Multiple Values)")
    print("7. Exit Program")

    choice = int(input("Please enter your choice: "))

    if choice == 1:
        print("\nStep 1: Input Data")
        data_list = input_data()
        print("\nData has been stored successfully!")

    elif choice == 2:
        print("\nStep 2: Display Data Summary (Built-in Functions)")
        display_summary(data_list)

    elif choice == 3:
        print("\nStep 3: Calculate Factorial (Recursion)")

        number = int(input("\nEnter a number to calculate its factorial: "))

        if number < 0:
            print("\nFactorial does not exist for negative numbers.")
        else:
            print("\nFactorial of", number, "is:", factorial(number))

    elif choice == 4:
        print("\nStep 4: Filter Data by Threshold (Lambda Function)")
        filter_data(data_list)

    elif choice == 5:
        print("\nStep 5: Sort Data")
        sort_data(data_list)

    elif choice == 6:
        print("\nStep 6: Display Dataset Statistics (Return Multiple Values)")

        stats = dataset_statistics(data_list)

        if stats:
            minimum, maximum, total, average = stats

            print("\nDataset Statistics:")
            print("- Minimum value:", minimum)
            print("- Maximum value:", maximum)
            print("- Sum of all values:", total)
            print("- Average value:", average)

    elif choice == 7:
        print("\nThank you for using the Data Analyzer and Transformer Program. Goodbye!")
        break

    else:
        print("\nInvalid choice! Please enter a number between 1 and 7.")
```
